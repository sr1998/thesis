```python
%load_ext autoreload
%autoreload 2
import pandas as pd

k_shot = 10
n_outer_splits = 10
n_inner_splits = 10
balanced_or_unbalanced = "unbalanced"

abundance_data = pd.read_csv(
    "../../data/sun_et_al_data/mpa4_species_profile_preprocessed.csv",
    index_col=0,
    header=0,
)
metadata = pd.read_csv(
    "../../data/sun_et_al_data/sample_group_species_preprocessed2.csv",
    index_col=0,
    header=0,
)  # 2 is for metadata with all columns
```

```python
from loguru import logger
import numpy as np
import json
import os
import sys

sys.path.append("../..")

import src.data.dataloader as dataloader
from src.global_vars import BASE_DATA_DIR


study_names: list = metadata["Project_1"].unique().tolist()
assert len(study_names) == 36


def get_train_and_test_df_from_data(
    abundance_data: pd.DataFrame, metadata: pd.DataFrame, test_study: str
):
    test_metadata_df = metadata[metadata["Project_1"] == test_study]
    test_data_df = abundance_data.loc[test_metadata_df.index]
    train_metadata_df = metadata.drop(index=test_metadata_df.index)
    train_data_df = abundance_data.drop(index=test_data_df.index)

    return train_data_df, train_metadata_df, test_data_df, test_metadata_df


data_splits_for_all_studies = {}
for test_study in study_names:
    print(test_study)
    (
        cross_val_data_selection,
        train_data_df,
        train_metadata_df,
        test_data_df,
        test_metadata_df,
    ) = dataloader.get_cross_validation_sun_et_al_data_splits(
        abundance_data,
        metadata,
        test_study=test_study,
        k_shot=k_shot,
        balanced_or_unbalanced=balanced_or_unbalanced,
        n_outer_splits=n_outer_splits,
        n_inner_splits=n_inner_splits,
    )

    print(cross_val_data_selection)
    display(test_data_df)
    raise SystemExit()

# create folder if non-existent
save_in = BASE_DATA_DIR / "sun_et_al_data" / "pre_selected_data_splits"
os.makedirs(save_in, exist_ok=True)

with open(save_in / f"AAA{k_shot}shot_{balanced_or_unbalanced}.json", "w") as f:
    json.dump(data_splits_for_all_studies, f, indent=4)

logger.info(
    f"Saved data splits for all studies in {save_in / f'{k_shot}_{balanced_or_unbalanced}.json'}"
)
logger.info(f"Finished creating data splits for all studies")

# some checks
```
